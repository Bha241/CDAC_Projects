```python
import os
import time
import requests

from PIL import Image
from io import BytesIO

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

# --------------------------
# CONFIG
# --------------------------

SAVE_DIR = "pinterest_pose_dataset"
QUERY = "cafe poses men"

MAX_IMAGES = 200

os.makedirs(SAVE_DIR, exist_ok=True)

# --------------------------
# SELENIUM DRIVER
# --------------------------

options = webdriver.ChromeOptions()

options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

# --------------------------
# OPEN PINTEREST
# --------------------------

url = f"https://www.pinterest.com/search/pins/?q={QUERY}"

driver.get(url)

time.sleep(5)

# --------------------------
# SCROLL TO LOAD IMAGES
# --------------------------

for _ in range(10):
    driver.execute_script(
        "window.scrollTo(0, document.body.scrollHeight);"
    )
    time.sleep(3)

# --------------------------
# EXTRACT IMAGE URLS
# --------------------------

images = driver.find_elements(By.TAG_NAME, "img")

img_urls = set()

for img in images:
    src = img.get_attribute("src")

    if src and "236x" in src:
        high_res = src.replace("236x", "originals")
        img_urls.add(high_res)

print(f"Found {len(img_urls)} image URLs")

# --------------------------
# DOWNLOAD IMAGES
# --------------------------

count = 0

for url in img_urls:

    if count >= MAX_IMAGES:
        break

    try:
        response = requests.get(url, timeout=15)

        image = Image.open(BytesIO(response.content)).convert("RGB")

        image = image.resize((224, 224))

        filename = f"pose_{count}.jpg"

        image.save(os.path.join(SAVE_DIR, filename))

        print(f"Saved: {filename}")

        count += 1

    except Exception as e:
        print("Error:", e)

driver.quit()

print(f"\nDownloaded {count} images")
```

```python
import requests
import os
import time
from PIL import Image
from io import BytesIO

API_KEY = "xP0D8DPPjYyfk5Bv8unmKMId8LKMxThJwqaV3DmMtJkMi302DzawFe6C"
SAVE_DIR = "pexels_cafe_pose_dataset23055"
TARGET = 1000

os.makedirs(SAVE_DIR, exist_ok=True)
headers = {"Authorization": API_KEY}

QUERIES = [
    "people sitting cafe modern",
    "coffee shop portrait young",
    "starbucks cafe person",
    "cafe couple sitting",
    "coffee shop lifestyle",
    "barista coffee shop",
    "cafe window seat person",
    "young adult coffee drink",
    "cafe friends talking",
    "laptop coffee shop working"
]

count = 0
seen_ids = set()

for QUERY in QUERIES:
    if count >= TARGET:
        break

    print(f"\nSearching: '{QUERY}'")
    page = 1

    while count < TARGET:
        params = {
            "query": QUERY,
            "per_page": 80,   # max allowed per page
            "page": page,
            "size": "medium",
            "orientation": "portrait"
        }

        response = requests.get(
            "https://api.pexels.com/v1/search",
            headers=headers,
            params=params
        )

        if response.status_code == 429:
            print("Rate limited — waiting 60s...")
            time.sleep(60)
            continue

        if response.status_code != 200:
            print(f"API error: {response.status_code}")
            break

        photos = response.json().get("photos", [])
        if not photos:
            print(f"No more results for '{QUERY}'")
            break

        print(f"  Page {page} — {len(photos)} photos found")

        for photo in photos:
            if count >= TARGET:
                break
            if photo["id"] in seen_ids:
                continue

            try:
                img_url = photo["src"]["large2x"]
                img_response = requests.get(img_url, timeout=15)
                img_response.raise_for_status()

                image = Image.open(BytesIO(img_response.content)).convert("RGB")
                image = image.resize((224, 224))

                filename = f"pexels_cafe_{photo['id']}.jpg"
                filepath = os.path.join(SAVE_DIR, filename)
                image.save(filepath)

                seen_ids.add(photo["id"])
                count += 1
                print(f"  [{count}/{TARGET}] Saved: {filename}")

            except Exception as e:
                print(f"  Error: {e}")

        page += 1
        time.sleep(0.5)  # be polite to API

print(f"\nDone. Total {count} images → {SAVE_DIR}/")
```
